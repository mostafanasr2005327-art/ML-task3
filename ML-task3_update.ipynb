## 5. Feature Engineering

Create an age-group feature from the existing age column to provide an additional input feature.

```python
# 1. Create a copy
df = df.copy()

# 2. Create age groups
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 25, 35, 50, 100],
    labels=['young', 'adult', 'middle_age', 'older']
)

# 3. Net capital
df['capital_net'] = (
    df['capital.gain'] - df['capital.loss']
)

# 4. Has capital gain or not
df['has_capital_gain'] = (
    df['capital.gain'] > 0
).astype(int)

# 5. Categorize working hours
df['hours_category'] = pd.cut(
    df['hours.per.week'],
    bins=[0, 20, 40, 60, float('inf')],
    labels=['low', 'normal', 'high', 'very_high']
)

# Check the new features
print(df[['age_group', 'capital_net',
          'has_capital_gain', 'hours_category']].head())

print(df.shape)
```

```python
# Separate features and target
X = df.drop("income", axis=1)
y = df["income"]

# Encode the target labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)

print(X.shape)
print(pd.Series(y).value_counts())
```

## 7. Import Pipeline Tools

## 8. Split Training and Test Data

Split the dataset into training and test sets. Stratification helps preserve the target-class proportions in both sets.

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
```

## 9. Build the Preprocessing Steps

Identify numeric and categorical columns. Fill missing numeric values with the median and scale them; fill missing categorical values with the most frequent value and encode them.

### Markdown Notes — Outlier Handling

- The IQR capper now calculates limits from numeric columns only.
- During `transform()`, only numeric columns are capped; categorical columns remain unchanged.
- Lower and upper outliers are checked independently, so both sides can be reported.


```python
X.skew(numeric_only=True)
```

```python
class IQRcapper(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        self.numeric_features = X.select_dtypes(include=np.number).columns
        X_numeric = X[self.numeric_features]

        Q1 = X_numeric.quantile(0.25)
        Q3 = X_numeric.quantile(0.75)
        IQR = Q3 - Q1

        self.lower = Q1 - 1.5 * IQR
        self.upper = Q3 + 1.5 * IQR

        return self

    def outlier(self, X):
        X = pd.DataFrame(X)
        for col in self.upper.index:
            outliers_lower = X[col] < self.lower[col]
            outliers_upper = X[col] > self.upper[col]

            if outliers_lower.any():
                print(f"outliers below lower bound in column: {col}")
                print(X.loc[outliers_lower, col])

            if outliers_upper.any():
                print(f"outliers above upper bound in column: {col}")
                print(X.loc[outliers_upper, col])

            if not outliers_lower.any() and not outliers_upper.any():
                print(f"no outliers in column: {col}")

    def transform(self, X):
        X = pd.DataFrame(X).copy()
        numeric_columns = [col for col in self.numeric_features if col in X.columns]

        X.loc[:, numeric_columns] = X.loc[:, numeric_columns].clip(
            lower=self.lower[numeric_columns],
            upper=self.upper[numeric_columns],
            axis=1
        )

        return X
```

### Markdown Notes — Preprocessing

- Numeric, categorical, and skewed columns are assigned to separate pipelines.
- Columns with `category` dtype are now included in categorical encoding.
- `capital.gain` and `capital.loss` receive `log1p` because they are non-negative and typically right-skewed.
- The skewed columns are excluded from the standard numeric pipeline to avoid processing the same feature twice.
- The preprocessing remains inside the model pipeline, so transformations are fitted on each training fold only.


```python
# Select numeric and categorical columns
skew_features = [col for col in ["capital.gain", "capital.loss" , "capital_net" , "has_capital_gain"] if col in X.columns]
numeric_features = X.select_dtypes(include=np.number).columns.difference(skew_features)
categorical_features = X.select_dtypes(include=["object", "category"]).columns

# Clean and scale numeric columns
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Clean and encode categorical columns
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Correct skewness only for selected non-negative, highly skewed columns
skew_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("skew", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),
    ("scaler", StandardScaler())
])

# Apply each preprocessing pipeline to its own columns
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features),
    ("skew", skew_pipeline, skew_features)
])
```

```python
capper = IQRcapper()

capper.fit(X)
capper.outlier(X)
```

## 10. Train and Tune the Model

Combine preprocessing, feature selection, and a Random Forest classifier into one pipeline. `GridSearchCV` tests the selected parameter combinations using cross-validation.

```python
# Build the machine learning pipeline
pipeline = Pipeline([
    ("outlier" , IQRcapper()),
    ("preprocessing", preprocessor),
    ("feature_selection", SelectKBest(score_func=mutual_info_classif)),
    ("smote", SMOTE(random_state=42)),
    ("model", RandomForestClassifier(random_state=42))
])

# Try different settings and choose the best one
param_grid = {
    "feature_selection__k": [10, "all"],
    "model__n_estimators": [100, 200],
    "model__max_depth": [None, 10]
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=3,
    scoring="accuracy",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best cross-validation score:", grid_search.best_score_)
```

## 11. Evaluate the Best Model

Use the best model found by cross-validation to predict the held-out test set, then report accuracy, classification metrics, and the confusion matrix.

```python
# Evaluate the best model on the test data
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Test accuracy:", accuracy)
print("\\nClassification report:")
print(classification_report(y_test, y_pred))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

Create a small table comparing the best cross-validation accuracy with the final test accuracy.

```python
# Track model performance
results = pd.DataFrame({
    "Metric": ["Best CV Accuracy", "Test Accuracy"],
    "Score": [grid_search.best_score_, accuracy]
})

results
```
